```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
data=pd.read_csv("kidney_disease.csv")
df=pd.DataFrame(data)
df
```

```python
df.drop("id",axis=1,inplace=True)
```

```python
df.columns=["Age","Blood Pressure","Specific Gravity","Albumin","Sugar","Red Blood Cells"," Pus Cells",
 "Pus Cell Clumps","Bacteria","Blood Glucose Random","Blood Urea","Serum Creatinine","Sodium",
 "Potassium","Hemoglobin","Packed Cell Volume"," White Blood Cell Count","Red Blood Cell Count",
 "Hypertension","Diabetes Mellitus","Coronary Artery Disease","Appetite","Pedal Edema","Anemia","Class"]
```

```python
df.info()
```

```python
# Now Convert Object columns which tends to be in numeric into numeric columns
#df["Packed Cell Volume"]=df["Packed Cell Volume"].astype(int)

# this method can't executed coz of null values
df["Packed Cell Volume"]=pd.to_numeric(df["Packed Cell Volume"],errors='coerce')
```

```python
df["Packed Cell Volume"].dtype
```

```python
df["Red Blood Cell Count"]=pd.to_numeric(df["Red Blood Cell Count"],errors='coerce')
df[" White Blood Cell Count"]=pd.to_numeric(df[" White Blood Cell Count"],errors='coerce')
```

```python
df["Red Blood Cell Count"].dtype
df[" White Blood Cell Count"].dtype
```

```python
cat_cols=[col for col in df.columns if df[col].dtype=="object"]
print(cat_cols)
num_cols=[col for col in df.columns if df[col].dtype!="object"]
print(num_cols)
```

```python
cat_cols
```

```python
for col in cat_cols:
    print(f"{col} has {df[col].unique()} values \n")
```

```python
# Now replace the mistakenly enteres data to new form
df["Diabetes Mellitus"].replace(to_replace={' yes':'yes','\tyes':'yes','\tno':'no'},inplace=True)
```

```python
df["Diabetes Mellitus"].unique()
```

```python
df["Coronary Artery Disease"].replace(to_replace={"\tno":"no"},inplace=True)
```

```python
df["Coronary Artery Disease"].unique()
```

```python
df["Class"].replace(to_replace={"ckd\t":"ckd"},inplace=True)
```

```python
df["Class"].unique()
```

```python
# Univariate Analysis >> Analysis of One Feature
plt.figure(figsize=(8,6))
sns.displot(df["Age"].dropna(),kde=True)
plt.title("Distribution Of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Insights >> The Age column is right Skewed
```
